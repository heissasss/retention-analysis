# Задача №3. График Retention

- **Столбцы** — количество договоров, закрытых за день.
- **Линии** — доля клиентов, которые взяли новый заём через 0, 1, 3 и 30 дней после закрытия предыдущего договора.

## Главный вывод

Retention вырос примерно до 88% за 30 дней, а количество закрытий увеличилось почти в 2,4 раза. 
Большинство клиентов возвращается за новым займом в первые 1–3 дня после закрытия предыдущего.

![](retention.png)


## Выводы

### 1. Количество закрытых договоров растёт

За период с 2 по 20 сентября количество закрытий увеличилось примерно с 4,5 до 11 тысяч в день. Это почти в 2,4 раза. 21 сентября показатель составил около 10 тысяч.

При таком росте объёма даже небольшое изменение Retention становится существенным.

### 2. Большинство клиентов возвращается в первые дни

Для закрытий за 13–16 сентября показатели составляют:

- 0 дней — около 37%;
- 1 день — около 63%;
- 3 дня — около 72%;
- 30 дней — около 88%.

Самый большой прирост происходит в первый день: примерно +26 п.п. Следующие два дня добавляют ещё около 9 п.п., а оставшиеся 27 дней — около 15 п.п. - наиболее перспективное время для предложения нового займа — первые 1–3 дня после закрытия предыдущего.

При этом около 12% клиентов не возвращаются даже в течение 30 дней.

### 3. Каждые семь дней наблюдаются резкие провалы

| Дата | 0 дней | 1 день |
|---|---:|---:|
| 4 сентября | ≈1% | ≈36% |
| 11 сентября | ≈1% | ≈31% |
| 18 сентября | ≈6% | ≈40% |
| Обычный уровень | ≈37% | ≈63% |

Возможно, в эти дни происходило что-то системное (временно недоступны выдачи, тех.работы)

## Итог

Динамика выглядит положительной: количество закрытий растёт, а доля клиентов, которые возвращаются за новым займом, достигает примерно 88% за 30 дней.

При этом основная часть возвратов происходит в первые три дня - это можно использовать при планировании коммуникаций с клиентами.
